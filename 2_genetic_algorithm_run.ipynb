# Genetic Algorithm Running
This notebook run the genetic algorithms and saves the outputs.

```python
%load_ext autoreload
%autoreload 2
```

```python
from utils.environment import Environment
from utils.notebook import ga_tuning_function

import pickle
```

```python
for city_name in ['grid_city', 'bottleneck_city', 'moderate_city']:
    city = Environment(city_name=city_name)
    n_experiments = 20
    num_agents = 20
    params = {"congestion": True, "walking": False, "fitness": "max"}
    population_size = 50

    hyperparams = {
        "crossover": 0.8,
        "mutation": 0.1,
        "epsilon": 0.2,
        "max_evolutions": 50,
        "survivor_method": 'elite_percentage',
        "tournament_size": None
    }

    solutions, evolution_stats, timings, seeds = ga_tuning_function(
        city, n_experiments, num_agents, params, population_size, hyperparams
    )
    
    path = f'outputs/{city_name}_{n_experiments}_experiments_analysis.pkl'
    with open(path, "wb") as f:
        results = {
            # results
            'solutions': solutions, 'evolution_stats': evolution_stats, 'timings': timings,
            # metadata
            'num_agents': num_agents, 'hyperparams': hyperparams, 
            'population_size': population_size, 'n_experiments': n_experiments,
            'population_creation_seed': seeds
        }
        pickle.dump(results, f)
    
```

## Running different selection approaches

```python
for city_name in ['grid_city', 'bottleneck_city', 'moderate_city']:
    for selection_method in ['elite_percentage', 'elite', 'children']:
        city = Environment(city_name=city_name)
        n_experiments = 20
        num_agents = 20
        params = {"congestion": True, "walking": False, "fitness": "max"}
        population_size = 20

        hyperparams = {
            "crossover": 0.8,
            "mutation": 0.1,
            "epsilon": 0.2,
            "max_evolutions": 50,
            "survivor_method": selection_method
        }

        solutions, evolution_stats, timings, seeds = ga_tuning_function(
            city, n_experiments, num_agents, params, population_size, hyperparams
        )
        
        path = f'outputs/{city_name}_{n_experiments}_survivor_selection_{selection_method}_experiments_analysis.pkl'
        print(path)
        with open(path, "wb") as f:
            results = {
                # results
                'solutions': solutions, 'evolution_stats': evolution_stats, 'timings': timings,
                # metadata
                'num_agents': num_agents, 'hyperparams': hyperparams, 
                'population_size': population_size, 'n_experiments': n_experiments,
                'population_creation_seed': seeds
            }
            pickle.dump(results, f)
    
```
