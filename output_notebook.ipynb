## Environment Set Up

```python
from utils.environment import Environment 
from utils.evaluation_metrics import FinalEvaluationMetrics

from multi_agent_ga import simulate_evacuation, greedy_algorithm

city = Environment(city_name="super_small_city_graph")
```

## Basic Genetic Algorithm

```python
solution_basic, evaluation_basic = simulate_evacuation(
    city=city,
    num_agents=10,
    population_size=50,
    algorithm_seed=29,
    max_evolutions=10,
    simulation_params={"congestion": True, "walking": True, "panic": False},
)
```

## Human Inspired Genetic Algorithm

```python
human_ga_params = {"congestion": True, "walking": True, "panic": True}
solution_human, evaluation_human = simulate_evacuation(
    city=city,
    num_agents=10,
    population_size=50,
    algorithm_seed=29,
    max_evolutions=20,
    simulation_params=human_ga_params,
)
```

## Greedy Algorithm

```python
greedy_params = {"congestion": True, "walking": True, "panic": True}
greedy_solution = greedy_algorithm(
    city=city,
    num_agents=10,
    simulation_params=greedy_params,
)
```

```python
final_greedy = FinalEvaluationMetrics(solution=greedy_solution, chromosome=False, params=greedy_params)
final_greedy.score()
```

```python
final_human = FinalEvaluationMetrics(solution=solution_human, chromosome=True, params=human_ga_params)
final_human.score()
```
